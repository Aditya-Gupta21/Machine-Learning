```python
# Importing the numpy 
import numpy as np
```

```python
# Creating the 0D tensors --> scaler 
a = np.array(19)
a
```

```python
#  here it the 0 dimensional array 
print(a.ndim)
```

```python
# creating the 1D tensor --> vector 
b = np.array([1,2,3,4,5])
b
```

```python
b.ndim
```

```python
# creating an 2d tensor --> matrice 
c = np.arange(1,10).reshape(3,3)
c
```

```python
c.ndim
```

```python
# creating an 3d tensor 
d = np.arange(1,28).reshape(3,3,3)
d
```

```python
d.ndim
```

```python
# creating an 4d tensors 
arr = np.arange(1,17).reshape(2,2,2,2)
arr
```
