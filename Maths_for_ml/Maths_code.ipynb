```python
# Importing the library of the python 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# here the array is 
arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
arr 
```

```python
# here we have to calculate the q1 and q3 
Q1 = np.percentile(arr,25)
Q3 = np.percentile(arr,75)
```

```python
# here e have to calculate the IQR 
IQR = Q3 - Q1
IQR
```

```python
# here we have to calculate the upper fence and lower fence 
UF = Q3 + (1.5 * (IQR))
LF = Q1 - (1.5*(IQR))
```

```python
# Use the for loop to print the element 
l = []
for i in arr :
    if i <= UF and i >= LF:
       l.append(i)

arr2 = np.array(l)
```
