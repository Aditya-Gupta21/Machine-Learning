PROBABLITY MASS FUNCTION 

```python
import pandas as pd 
import random
```

```python
# making an empty list 
L = []
for i in range(100000):
    a = random.randint(1,6)
    b = random.randint(1,6)

    L.append(a+b)
```

```python
len(L)
```

```python
print(L[:6])
```

```python
s = (pd.Series(L).value_counts()/pd.Series(L).value_counts().sum()).sort_index()
s
```

```python
# plooting 
s.plot(kind = 'bar')
```

CUMMULATIVE DISTRIBUTION FUNCTION (CDF) USING PMF

```python
import numpy as np

# cummulative frequency values always beetween 0 and 1 
np.cumsum(s)
```

```python
np.cumsum(s).plot(kind = 'bar')
```

PROBABILITY DENSITY FUNCTION

```python
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
from numpy.random import normal
```

```python
sample = normal(loc = 50,scale = 5,size = 1000)
sample
```

```python
#  plotting the histrogram of the sample data 
plt.hist(sample,bins = 20)
```

```python
#  from the sample data we have to calculate mean and standard deviation 
sample_mean = sample.mean()
sample_std = sample.std()
```

```python
from scipy.stats import norm
dist = norm(sample_mean,sample_std)
```

```python
values = np.linspace(sample.min(),sample.max(),100)
```

```python
probability = [dist.pdf(value) for value in values]
```

```python
plt.hist(sample,bins = 10,density = True)
plt.plot(values,probability)
```
