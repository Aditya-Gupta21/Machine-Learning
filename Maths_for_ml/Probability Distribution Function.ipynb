PROBABLITY MASS FUNCTION 

```python
import pandas as pd 
import random
```

```python
# making an empty list 
L = []
for i in range(100000):
    a = random.randint(1,6)
    b = random.randint(1,6)

    L.append(a+b)
```

```python
len(L)
```

```python
print(L[:6])
```

```python
s = (pd.Series(L).value_counts()/pd.Series(L).value_counts().sum()).sort_index()
s
```

```python
# plooting 
s.plot(kind = 'bar')
```
