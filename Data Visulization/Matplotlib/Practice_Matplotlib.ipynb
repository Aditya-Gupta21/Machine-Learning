```python
#  Import the library of the python 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
#  Creating the first array 
x = np.random.random(15)
x

# Creating the second array 
y = np.arange(1,16)
y
```

```python
#  Plotting the graph 
plt.plot(x,y)

# Setting the title 
plt.title("Graph")

# Setting the x Axis 
plt.xlabel("X Axis")

# Setting the Y Axis 
plt.ylabel("Y Axis")
```

PRACTISING THE SUBPLOT

```python
#  Making the subplot of the of 2,2 
# Here i print the values in the subplot 

#  Printing the plot in the subplot 1
plt.subplot(2,2,1)    # here it print the (0,0) of the subplot
plt.plot(x,y)

# Printing the plot in the subplot 2
plt.subplot(2,2,3)    # here it print the (1,0) of the subplot
plt.plot(x/2,y**7)

# printing the plot in the subplot 3 
plt.subplot(2,2,2)     # here it print the (0,1) of the subplot
plt.plot(y,x)

# printing the values in the subplot 4 
plt.subplot(2,2,4)     # here it print the (1,1) of the subplot
plt.plot(x**4,y**1)
```

Practising the figure 

```python
fig = plt.figure()

axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)

axis2 = fig.add_axes([0.3,0.4,0.5,0.5])
axis2.plot(y,x)
```

TYPES OF THE PLOTS 

```python
#  1 --> Scatter plot 
plt.scatter(x,y)
```

```python
#  Hist plot 
plt.hist(x)
```

SAVING AN IMAGE 

```python
fig = plt.figure(dpi = 200)
axis1 = fig.add_axes([0.1,0.1,0.7,0.5])
axis1.plot(y,x)
fig.savefig("saving_img.png")   # here we have to save the figure not axis 
```
