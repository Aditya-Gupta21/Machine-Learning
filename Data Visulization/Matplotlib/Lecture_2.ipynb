```python
# Import all library of the python 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
#  Making the nd array 
x = np.linspace(0,3,10)
x

# Making the nd array 
y = x**2
y
```

```python
#  Learn the subplot 
# when we make we have to select which one plot 
# When we have to make the plot in the subplot then we right the ploting code down to the subplot 

# Making the plot inside the subplot 1 
plt.subplot(2,2,1)   # Here we select the first subplot 
plt.plot(x,y)

# Making the plot inside the subplot 2
plt.subplot(2,2,2)  # Here we select the second subplot 
plt.plot(y,x)

# Making the plot inside the subplot 3
plt.subplot(2,2,3)  # here we select the third subplot 
plt.plot(y**3,x**2)


# Making the plot inside the subplot 4
plt.subplot(2,2,4)   #Here we select the fourth subplot
plt.plot(x**4,y**2)
```

```python
#  LEARNING OF OBJECT ORIENTED WAY 
fig = plt.figure(figsize = (4,6),dpi = 100)

# Makin the axis1 
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)

# Making the axis 2 
axis2  = fig.add_axes([0.4,0.3,.5,.5])
axis2.plot(x**3,y)
```

TYPES OF THE PLOTS

1 --> Scatter plot --> used in regression algorithm

```python
plt.scatter(x,y)
```

2 --> HISTROGRAM PLOT --> in form of histrogram 

```python
plt.hist(y)
```

SAVING AN IMAGE 

```python
fig = plt.figure()
axis = fig.add_axes([0.1,0.1,0.8,0.8])
axis.plot(x,y)
plt.savefig("Basic_img.png")
```
