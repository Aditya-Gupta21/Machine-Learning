```python
#  Import the matplotlib and start learning from start 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
```

```python
y = x**2
y
```

```python
plt.plot(x,y)

# If we want title to our plot 
plt.title('First Plot')

# If we want label at x axis 
plt.xlabel("X axis")

# If we want label at y axis 
plt.ylabel("Y axis")
```
