```python
#  Import the following library of the python 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

DISTRIBUTION PLOT --> It create the distribution of quantitative data 

Types of the distribution plot
1. DISTPLOT/HISTPLOT
2. JOINTPLOT
3. PAIR PLOT
4. RUGPLOT

```python
#  We have to load dataset from the github of the seaborn 
df = sns.load_dataset("tips")
df 
```

DISPLOT/HISTPLOT

```python
df['size'].unique()
```

```python
#  Making the plot 

# Making the plot in the 1 row and 2 columns 
# making the first plot whose axis are (row = 0 anf column = 0)
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins = 30,kde = True)

# Making the second plot whose axis are (row = 0 and columns = 1)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins = 30,kde = True)
```

2. JOINTPLOT

```python
# Mkaing the joint plot 
df
```

```python
sns.jointplot(x = 'total_bill',y = 'tip',data = df,kind = 'scatter')
```

```python
sns.jointplot(x = 'total_bill',y = 'tip',data = df,kind = 'kde')
```

```python
sns.jointplot(x = 'total_bill',y = 'tip',data = df,kind = 'reg')
```

```python
sns.jointplot(x = 'total_bill',y = 'tip',data = df,kind = 'hex')
```

```python
sns.jointplot(x = 'total_bill',y = 'tip',data = df,kind = 'resid')
```

3. Pair plot

```python
sns.pairplot(df,hue = 'sex')
```

```python
sns.rugplot(df)
```
