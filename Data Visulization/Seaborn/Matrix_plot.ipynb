```python
# Import the library of the python 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

HEAT MAP

```python
#  loading a data set 
flights = sns.load_dataset("flights")
flights
```

```python
#  Loading the tips dataset 
tips = sns.load_dataset("tips")
tips
```

```python
#  it shuould be negative value 
tipsCorr = tips[['total_bill','tip','size']]
tipsCorr
```

```python
tipsCorr.corr()
```

```python
sns.heatmap(tipsCorr.corr())
```

CLUSTER MAP 

```python
sns.clustermap(tipsCorr.corr())
```

pivot table and heat map

```python
pvtflights = flights.pivot_table(values = 'passengers',index = 'month',columns ='year' )
pvtflights
```

```python
sns.heatmap(pvtflights)
```
