```python
#  Learn about the categorical plot 
import numpy as np
import pandas as pd 
import matplotlib.pyplot as pd 
import seaborn as sns 
```

CATGORICAL PLOTS --> work the quanlitative data set 
Types of the categorical plots 
1. boxplot
2. violinplot
3. stripplot
4. swarmplot
5. barplot
6. countplot

```python
df = sns.load_dataset("tips")
df 
```

```python
sns.countplot(x = df['sex'],hue = df['smoker'])
```

```python
sns.barplot(x = df['sex'],y = df['total_bill'])
```

```python
sns.barplot(x = df['sex'],y = df['total_bill'],  estimator = np.max)
```

```python
sns.barplot(x = df['sex'],y = df['total_bill'],  estimator = np.min)
```

```python
sns.boxplot(x ='tip', y = 'day', data = df)
```

VIOLEIN PLOT

```python
sns.violinplot(x= 'day',y ='size',data = df )
```

STRIPLOT 

```python
sns.stripplot(x = 'day', y = 'tip',data = df)
```
